### Liczenie podfolerów i plików

```python
import os

ROOT_DIR = "C:\\Users\\kam45x\\OneDrive - Politechnika Warszawska\\przekroje"

valid_dirs = 0
valid_files = 0

for root, dirs, files in os.walk(ROOT_DIR):
    if not dirs:  # only deepest subdirectories
        # Convert to relative path
        rel = os.path.relpath(root, ROOT_DIR)

        # Check for I10/I10.dcm
        if any(f in ("I10", "I10.dcm") for f in files):
            valid_dirs += 1
            valid_files += sum(1 for f in files if f[0] == "I")

print(f"\nTotal directories to process: {valid_dirs}\n")
print(f"Total files to process: {valid_files}\n")
```

```python
DIR_LIST_FILE = "../../src/ct_scan_accepted.txt"

valid_dirs = 0
valid_files = 0

# Wczytaj listę katalogów
with open(DIR_LIST_FILE, "r") as f:
    dir_list = [line.strip() for line in f if line.strip()]

# Przetwarzanie katalogów
for rel_path in dir_list:
    abs_path = os.path.join(ROOT_DIR, rel_path)

    if not os.path.isdir(abs_path):
        print(f"Warning: directory does not exist: {abs_path}")
        continue

    files = os.listdir(abs_path)

    if any(f in ("I10", "I10.dcm") for f in files):
        valid_dirs += 1
        valid_files += sum(1 for f in files if f[0] == "I")

print(f"\nTotal directories to process: {valid_dirs}\n")
print(f"Total files to process: {valid_files}\n")
```
